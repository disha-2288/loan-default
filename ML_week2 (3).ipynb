```python
df = pd.get_dummies(df, columns=['Education'],drop_first = True)
df
#OneHotEncoder
```

```python
binary_cols = ["HasMortgage", "HasDependents", "HasCoSigner"]

for col in binary_cols:
    df[col] = df[col].map({
        "No": 0,
        "Yes": 1
    })
    
```

```python
df['Default'].value_counts().plot(kind='pie', autopct='%1.1f%%', colors=['g', 'b'], labels=['Paid Back (0)', 'Defaulted (1)'])
plt.title('Overall Default Distribution')
plt.ylabel('')
plt.show()
```

```python
#avg DIT per class
df.groupby('Default')['DTIRatio'].mean() 
```

```python
 # loan amount  Median
df.groupby('Default')['LoanAmount'].median()
```

```python
#Job Category V/S Default Rate
df.groupby('EmploymentType')['Default'].mean().sort_values().plot(kind='barh', color='orange', edgecolor='black')
plt.title('Default Rates by Employment Category')
plt.show()
```

```python
# avg Income by emp type 
avg_income_job = df.groupby('EmploymentType')['Income'].mean().sort_values()
avg_income_job.plot(kind='bar', color='skyblue', edgecolor='black' )
plt.title('Average Income Across Employment Categories')
plt.ylabel('Average Income ($)')
plt.xlabel('Employment Type')
plt.xticks(rotation=45)
plt.show()
```
